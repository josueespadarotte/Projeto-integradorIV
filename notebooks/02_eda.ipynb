```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('prf_es_clean.csv')

#Tratamento correto do horário
df['hora_inteira'] = pd.to_datetime(df['horario'], format='%H:%M:%S', errors='coerce').dt.hour

#Gerar o gráfico de Horários de Pico
sns.set_theme(style="whitegrid")
plt.figure(figsize=(12, 6))

sns.countplot(data=df, x='hora_inteira', hue='classificacao_acidente', palette='viridis')

plt.title('Distribuição de Acidentes por Horário (ES) - Sem Avisos')
plt.xlabel('Hora do Dia')
plt.ylabel('Quantidade de Acidentes')
plt.legend(title='Gravidade', loc='upper right')
plt.show()
```

```python
# Gráfico de Horários de Pico
plt.figure(figsize=(12, 6))

# Usamos countplot para contar acidentes por hora e mostrar a gravidade (hue)
sns.countplot(data=df, x='hora_inteira', hue='classificacao_acidente', palette='viridis')

plt.title('Distribuição de Acidentes por Horário e Gravidade (ES)')
plt.xlabel('Hora do Dia')
plt.ylabel('Quantidade de Acidentes')
plt.legend(title='Gravidade') # Agora a legenda terá dados para mostrar
plt.show()

# Rodovias mais Perigosas
plt.figure(figsize=(10, 5))
top_rodovias = df['br'].value_counts().head(5)

#hue=index para evitar o aviso anterior e legend=False porque não tem categorias extras
sns.barplot(x=top_rodovias.index, y=top_rodovias.values, hue=top_rodovias.index, palette='magma', legend=False)

plt.title('Top 5 Rodovias com Maior Volume de Acidentes')
plt.xlabel('Rodovia (BR)')
plt.ylabel('Total de Ocorrências')
plt.show()
```
